```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

• numpy digunakan untuk pengiraan nombor.
• pandas digunakan untuk mengurus data jadual.
• matplotlib digunakan untuk menghasilkan graf.
• np, pd dan plt ialah nama ringkas bagi setiap library.

```python
markah = np.array([70, 85, 90, 65, 80])
print(markah)
```

np.array() digunakan untuk membina satu array yang mengandungi beberapa nilai

```python
purata = np.mean(markah)
print("Purata markah:", purata)
```

```python
tertinggi = np.max(markah)
print("Markah tertinggi:", tertinggi)
```

```python
terendah = np.min(markah)
print("Markah terendah:", terendah)
```

```python
jumlah = np.sum(markah)
print("Jumlah markah:", jumlah)
```

```python
markah_baru = markah + 5
print(markah_baru)
```

Rumusan NumPy
NumPy boleh digunakan untuk:
• Mengira purata.
• Mencari nilai tertinggi.
• Mencari nilai terendah.
• Menjumlahkan data.
• Melakukan operasi matematik pada semua data serentak.


```python
data = {
"Nama": ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"],
"Markah": [70, 85, 90, 65, 80]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

DataFrame ialah jadual yang mempunyai baris dan lajur

```python
print(df.head())
```

```python
 print(df["Nama"])
```

```python
print(df["Markah"])
```

```python
purata = df["Markah"].mean()
print("Purata markah:", purata)
```

```python
pelajar_cemerlang = df[df["Markah"] >= 80]
print(pelajar_cemerlang)
```

```python
df["Status"] = np.where(df["Markah"] >= 50, "Lulus", "Gagal")
print(df)
```

```python
def tentukan_gred(markah):
    if markah >= 80:
        return "A"
    elif markah >= 65:
        return "B"
    elif markah >= 50:
        return "C"
    else:
        return "D"

df["Gred"] = df["Markah"].apply(tentukan_gred)
print(df)
```

```python
df.to_csv("markah_pelajar.csv",
index=False)
print("Fail berjaya disimpan.")
```

Fail akan disimpan dengan nama:
markah_pelajar.csv

```python
data_baru = pd.read_csv("markah_pelajar.csv")
print(data_baru)
```
