```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Nama": ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"],
    "Markah": [70, 85, 90, 65, 80]
}

df = pd.DataFrame(data)
purata = np.mean(df["Markah"])
tertinggi = np.max(df["Markah"])
terendah = np.min(df["Markah"])

print("Purata:", purata)
print("Markah tertinggi:", tertinggi)
print("Markah terendah:", terendah)
df["Status"] = np.where(df["Markah"] >= 50, "Lulus", "Gagal")

print(df)
plt.bar(df["Nama"], df["Markah"])

plt.axhline(
    y=purata,
    linestyle="--",
    label="Purata"
)

plt.title("Analisis Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")
plt.legend()

plt.show()
```

```python
plt.bar(df["Nama"], df["Markah"])

plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

plt.show()
```

GRAF BAR MARKAH PELAJAR

```python
plt.plot(df["Nama"], df["Markah"], marker="o")

plt.title("Prestasi Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

plt.show()
```

GRAF GARISAN

```python
plt.hist(df["Markah"], bins=5)

plt.title("Taburan Markah Pelajar")
plt.xlabel("Julat Markah")
plt.ylabel("Bilangan Pelajar")

plt.show()
```

HISTOGRAM

```python
# 1. Cipta fungsi untuk tentukan gred berdasarkan markah
def bina_gred(markah):
    if markah >= 85:
        return 'A'
    elif markah >= 70:
        return 'B'
    elif markah >= 50:
        return 'C'
    else:
        return 'F'

# 2. Tambah lajur 'Gred' ke dalam DataFrame
df["Gred"] = df["Markah"].apply(bina_gred)

# 3. Jalankan kod carta pai anda semula
jumlah_gred = df["Gred"].value_counts()

plt.pie(
    jumlah_gred,
    labels=jumlah_gred.index,
    autopct="%1.1f%%"
)

plt.title("Peratusan Gred Pelajar")
plt.show()
```
